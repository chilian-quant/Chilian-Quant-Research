```python
# 1. Instalar la librería (solo la primera vez)
!pip install yfinance

# 2. Importar librerías
import yfinance as yf
import matplotlib.pyplot as plt

# 3. Definir el ticker (Ejemplo: S&P 500)
ticker = "SPY"

# 4. Descargar datos históricos (últimos 5 años)
data = yf.download(ticker, period="5y")

# 5. Calcular volatilidad y graficar
data['Close'].plot(title=f"Chilian Quant - {ticker} Price Analysis", figsize=(10, 5))
plt.ylabel("Price (USD)")
plt.show()
```
